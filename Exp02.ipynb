```python
# ============================================================
# EXPERIMENT 2B: DIFFERENT MARKER STYLES
# ============================================================
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import warnings
from matplotlib import MatplotlibDeprecationWarning

warnings.filterwarnings("ignore", category=MatplotlibDeprecationWarning)

# --- Generate and Save Dataset ---
np.random.seed(42)
n = 100
study_hours = np.round(np.random.uniform(1, 10, n), 1)
marks = np.clip(np.round(30 + 6 * study_hours + np.random.normal(0, 5, n)), 0, 100).astype(int)
attendance = np.clip(np.round(60 + 3.5 * study_hours + np.random.normal(0, 4, n)), 50, 100).astype(int)

data = pd.DataFrame({
    "Name": [f"Student_{i+1}" for i in range(n)],
    "Department": np.random.choice(["CSE", "IT", "ECE"], n),
    "Marks": marks, 
    "Attendance (%)": attendance, 
    "Study_Hours": study_hours
})
data.to_csv("student_data.csv", index=False)
print(f"student_data.csv created with {len(data)} rows\n{data.head()}")

# --- Read Dataset and Display First Five Records ---
data = pd.read_csv("student_data.csv")
print(f"\nFirst Five Records\n{data.head()}")

# --- Plot Different Marker Styles ---
x, y = data["Study_Hours"], data["Marks"]
plt.figure(figsize=(10, 6))

# Define marker properties
markers = ['o', 's', '^', 'D', '*']
colors = ['red', 'blue', 'green', 'orange', 'purple']
labels = ['Circle', 'Square', 'Triangle', 'Diamond', 'Star']
sizes = [120, 120, 120, 120, 160]

# Loop through properties to plot each style
for i, (m, c, l, s) in enumerate(zip(markers, colors, labels, sizes)):
    plt.scatter(x[i*20:(i+1)*20], y[i*20:(i+1)*20], marker=m, s=s, color=c, label=l)

plt.title("Scatter Plot using Different Marker Styles")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)

# --- Save and Display Graph ---
plt.savefig("MarkerStyles.pdf")
plt.savefig("MarkerStyles.png")
plt.show()

print("\nMarker style chart created successfully.\nFiles Created:\n1. MarkerStyles.pdf\n2. MarkerStyles.png")
```

```python
# ============================================================
# EXPERIMENT 2H: MULTIPLE SCATTER PLOTS
# ============================================================
import pandas as pd
import matplotlib.pyplot as plt

# Read and Display Dataset
data = pd.read_csv("student_data.csv")
print(f"First Five Records\n{data.head()}")

# Divide Students into Categories
low = data[data["Marks"] < 60]
med = data[(data["Marks"] >= 60) & (data["Marks"] < 80)]
high = data[data["Marks"] >= 80]

# Create Figure and Plot Performers using a Loop
plt.figure(figsize=(8, 6))
for df, c, m, s, l in [(low, "red", "o", 120, "Marks < 60"), (med, "orange", "^", 120, "Marks 60-79"), (high, "green", "*", 170, "Marks >= 80")]:
    plt.scatter(df["Study_Hours"], df["Marks"], color=c, marker=m, s=s, label=l)

# Add Title, Labels, Legend, and Grid
plt.title("Multiple Scatter Plot")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)

# Save and Display Graph
plt.savefig("Multiple_Scatter_Plot.pdf")
plt.savefig("Multiple_Scatter_Plot.png")
plt.show()

# Display Success Message
print("Multiple scatter plot generated successfully.\n\nFiles Created:\n1. Multiple_Scatter_Plot.pdf\n2. Multiple_Scatter_Plot.png")
```

```python
# ============================================================
# EXPERIMENT 2I: MULTIPLE SUBPLOTS
# ============================================================
import pandas as pd
import matplotlib.pyplot as plt

# Read Dataset and Display First Five Records
data = pd.read_csv("student_data.csv")
print(f"First Five Records\n{data.head()}")
plt.figure(figsize=(12, 5))

# Subplot 1: Scatter Plot
plt.subplot(1, 2, 1)
plt.scatter(data["Study_Hours"], data["Marks"], color="royalblue", marker="o", s=100, edgecolors="black")
plt.title("Scatter Plot"), plt.xlabel("Study Hours"), plt.ylabel("Marks"), plt.grid(True)

# Subplot 2: Line Plot
plt.subplot(1, 2, 2)
plt.plot(data["Study_Hours"], data["Marks"], color="darkred", linewidth=2, marker="o", markersize=5)
plt.title("Line Plot"), plt.xlabel("Study Hours"), plt.ylabel("Marks"), plt.grid(True)

# Adjust Spacing, Save, and Display Graph
plt.tight_layout()
plt.savefig("Multiple_Subplots.pdf"), plt.savefig("Multiple_Subplots.png"), plt.show()
print("Multiple subplots generated successfully.\nFiles Created:\n1. Multiple_Subplots.pdf\n2. Multiple_Subplots.png")
```

```python
# ============================================================
# EXPERIMENT 2J: SCATTER PLOT WITH COLOR MAP
# ============================================================
import pandas as pd
import matplotlib.pyplot as plt

# Read Dataset and Display First Five Records
data = pd.read_csv("student_data.csv")
print(f"First Five Records\n{data.head()}")

# Create Scatter Plot Using Attendance as the Color Scale
plt.figure(figsize=(8, 6))
scatter = plt.scatter(data["Study_Hours"], data["Marks"], c=data["Attendance (%)"],
                      cmap="viridis", s=140, edgecolors="black")

# Add Color Bar, Title, Labels, and Grid
plt.colorbar(scatter, label="Attendance (%)")
plt.title("Scatter Plot with Color Map")
plt.xlabel("Study Hours"), plt.ylabel("Marks")
plt.grid(True, linestyle="--", alpha=0.5)

# Save and Display Graph
plt.savefig("Scatter_ColorMap.pdf"), plt.savefig("Scatter_ColorMap.png"), plt.show()
print("Scatter plot with color map generated successfully.\nFiles Created:\n1. Scatter_ColorMap.pdf\n2. Scatter_ColorMap.png")
```

```python
# ============================================================
# EXPERIMENT 2K: SCATTER PLOT WITH ANNOTATIONS
# ============================================================
import pandas as pd
import matplotlib.pyplot as plt

# Read Dataset and Display First Five Records
data = pd.read_csv("student_data.csv")
print(f"First Five Records\n{data.head()}")

# Create Scatter Plot
plt.figure(figsize=(9, 6))
plt.scatter(data["Study_Hours"], data["Marks"], color="royalblue", marker="o", s=120, edgecolors="black")

# Annotate First 10 Students
for i in range(10):
    plt.text(data["Study_Hours"].iloc[i] + 0.05, data["Marks"].iloc[i] + 0.5, data["Name"].iloc[i], fontsize=8)

# Add Title, Labels, and Grid
plt.title("Scatter Plot with Annotations")
plt.xlabel("Study Hours"), plt.ylabel("Marks")
plt.grid(True, linestyle="--", alpha=0.5)

# Save and Display Graph
plt.savefig("Scatter_Annotations.pdf"), plt.savefig("Scatter_Annotations.png"), plt.show()
print("Scatter plot with annotations generated successfully.\nFiles Created:\n1. Scatter_Annotations.pdf\n2. Scatter_Annotations.png")
```
