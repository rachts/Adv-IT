```python
# ============================================================
# EXPERIMENT 2B : DIFFERENT MARKER STYLES
# ============================================================
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt

import warnings
from matplotlib import MatplotlibDeprecationWarning
warnings.filterwarnings("ignore", category=MatplotlibDeprecationWarning)
# Recreate student_data.csv with Study_Hours and 100 records (run once)
import numpy as np
import pandas as pd

np.random.seed(42)
n = 100

study_hours = np.round(np.random.uniform(1, 10, n), 1)
marks = np.clip(np.round(30 + 6 * study_hours + np.random.normal(0, 5, n)), 0, 100).astype(int)
attendance = np.clip(np.round(60 + 3.5 * study_hours + np.random.normal(0, 4, n)), 50, 100).astype(int)

data = pd.DataFrame({
    "Name": [f"Student_{i+1}" for i in range(n)],
    "Department": np.random.choice(["CSE", "IT", "ECE"], n),
    "Marks": marks,
    "Attendance (%)": attendance,
    "Study_Hours": study_hours
})

data.to_csv("student_data.csv", index=False)
print("student_data.csv created with", len(data), "rows")
print(data.head())

# Read the dataset
data = pd.read_csv("student_data.csv")

# Display the first five records
print("First Five Records")
print(data.head())

# Select X and Y values
x = data["Study_Hours"]
y = data["Marks"]

# Create figure
plt.figure(figsize=(10, 6))

# Plot different marker styles
plt.scatter(x[:20], y[:20],
            marker='o',
            s=120,
            color='red',
            label='Circle')

plt.scatter(x[20:40], y[20:40],
            marker='s',
            s=120,
            color='blue',
            label='Square')

plt.scatter(x[40:60], y[40:60],
            marker='^',
            s=120,
            color='green',
            label='Triangle')

plt.scatter(x[60:80], y[60:80],
            marker='D',
            s=120,
            color='orange',
            label='Diamond')

plt.scatter(x[80:100], y[80:100],
            marker='*',
            s=160,
            color='purple',
            label='Star')

# Add chart title
plt.title("Scatter Plot using Different Marker Styles")

# Axis labels
plt.xlabel("Study Hours")
plt.ylabel("Marks")

# Display legend
plt.legend()

# Display grid
plt.grid(True, linestyle='--', alpha=0.5)

# Save the graph
plt.savefig("MarkerStyles.pdf")
plt.savefig("MarkerStyles.png")

# Display graph
plt.show()

print("Marker style chart created successfully.")

print("Files Created:")
print("1. MarkerStyles.pdf")
print("2. MarkerStyles.png")
```

```python
# ============================================================
# EXPERIMENT 2G : CUSTOMIZED LINE PLOT
# ============================================================
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt

# Read the dataset
data = pd.read_csv("student_data.csv")

# Display first five records
print("First Five Records")
print(data.head())

# Sort dataset by Study Hours
data = data.sort_values(by="Study_Hours")

# Create figure
plt.figure(figsize=(8, 6))

# Customized line plot
plt.plot(
    data["Study_Hours"],
    data["Marks"],
    linestyle="-",
    linewidth=3,
    marker="o",
    markersize=8,
    markerfacecolor="yellow",
    markeredgecolor="black",
    color="royalblue",
    label="Student Performance"
)

# Chart title
plt.title("Customized Line Plot")

# Axis labels
plt.xlabel("Study Hours")
plt.ylabel("Marks")

# Legend
plt.legend()

# Grid
plt.grid(True, linestyle="--", alpha=0.5)

# Save graph
plt.savefig("Customized_Line_Plot.pdf")
plt.savefig("Customized_Line_Plot.png")

# Display graph
plt.show()

print("Customized line plot generated successfully.")

print("Files Created:")
print("1. Customized_Line_Plot.pdf")
print("2. Customized_Line_Plot.png")
```

```python
# ============================================================
# EXPERIMENT 2H : MULTIPLE SCATTER PLOTS
# ============================================================
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt

# Read the dataset
data = pd.read_csv("student_data.csv")

# Display first five records
print("First Five Records")
print(data.head())

# Divide students into categories
low = data[data["Marks"] < 60]
medium = data[(data["Marks"] >= 60) & (data["Marks"] < 80)]
high = data[data["Marks"] >= 80]

# Create figure
plt.figure(figsize=(8, 6))

# Low performers
plt.scatter(
    low["Study_Hours"],
    low["Marks"],
    color="red",
    marker="o",
    s=120,
    label="Marks < 60"
)

# Medium performers
plt.scatter(
    medium["Study_Hours"],
    medium["Marks"],
    color="orange",
    marker="^",
    s=120,
    label="Marks 60-79"
)

# High performers
plt.scatter(
    high["Study_Hours"],
    high["Marks"],
    color="green",
    marker="*",
    s=170,
    label="Marks >= 80"
)

# Title
plt.title("Multiple Scatter Plot")

# Axis labels
plt.xlabel("Study Hours")
plt.ylabel("Marks")

# Legend
plt.legend()

# Grid
plt.grid(True, linestyle="--", alpha=0.5)

# Save graph
plt.savefig("Multiple_Scatter_Plot.pdf")
plt.savefig("Multiple_Scatter_Plot.png")

# Display graph
plt.show()

print("Multiple scatter plot generated successfully.")

print("Files Created:")
print("1. Multiple_Scatter_Plot.pdf")
print("2. Multiple_Scatter_Plot.png")
```
