```python
# ============================================================
# EXPERIMENT 3A : ADDING CHART TITLE AND AXIS LABELS
# ============================================================
import pandas as pd
import matplotlib.pyplot as plt

# Read dataset
df = pd.read_csv("student_data.csv")
df["Roll_No"] = range(1, len(df) + 1)

# Create figure
plt.figure(figsize=(8,5))

# Plot graph
plt.plot(
    df["Roll_No"],
    df["Marks"],
    marker="o"
)

# Add title
plt.title("Student Marks Analysis")

# Axis labels
plt.xlabel("Roll Number")
plt.ylabel("Marks")

# Display graph
plt.show()
```

```python
# ============================================================
# EXPERIMENT 3B : ADDING GRID LINES TO A GRAPH
# ============================================================
import pandas as pd
import matplotlib.pyplot as plt

# Read dataset
df = pd.read_csv("student_data.csv")
df["Roll_No"] = range(1, len(df) + 1)

# Create figure
plt.figure(figsize=(8,5))

# Plot attendance
plt.plot(
    df["Roll_No"],
    df["Attendance (%)"],
    marker="o"
)

# Add title
plt.title("Attendance Analysis")

# Axis labels
plt.xlabel("Roll Number")
plt.ylabel("Attendance (%)")

# Display grid
plt.grid(True)

# Display graph
plt.show()
```

```python
# ============================================================
# EXPERIMENT 3C : ANNOTATING IMPORTANT DATA POINTS
# ============================================================
import pandas as pd
import matplotlib.pyplot as plt

# Read dataset
df = pd.read_csv("student_data.csv")
df["Roll_No"] = range(1, len(df) + 1)

# Create figure
plt.figure(figsize=(8,5))

# Plot graph
plt.plot(
    df["Roll_No"],
    df["Marks"],
    marker="o"
)

# Find highest marks
highest = df["Marks"].idxmax()

# Add annotation
plt.annotate(
    "Highest Marks",
    xy=(
        df.loc[highest, "Roll_No"],
        df.loc[highest, "Marks"]
    ),
    xytext=(
        df.loc[highest, "Roll_No"] + 2,
        df.loc[highest, "Marks"] + 5
    ),
    arrowprops=dict(
        facecolor="red"
    )
)

# Add title
plt.title("Highest Marks")

# Display grid
plt.grid(True)

# Display graph
plt.show()
```

```python
# ============================================================
# EXPERIMENT 3D : ADDING MULTIPLE ANNOTATIONS TO A GRAPH
# ============================================================
import pandas as pd
import matplotlib.pyplot as plt

# Read dataset
df = pd.read_csv("student_data.csv")
df["Roll_No"] = range(1, len(df) + 1)

# Create figure
plt.figure(figsize=(8,5))

# Plot graph
plt.plot(
    df["Roll_No"],
    df["Marks"],
    marker="o"
)

# Find highest and lowest marks
high = df["Marks"].idxmax()
low = df["Marks"].idxmin()

# Highest annotation
plt.annotate(
    "Highest",
    xy=(df.loc[high, "Roll_No"],
        df.loc[high, "Marks"]),
    xytext=(df.loc[high, "Roll_No"] + 1,
        df.loc[high, "Marks"] + 5),
    arrowprops=dict(facecolor="green")
)

# Lowest annotation
plt.annotate(
    "Lowest",
    xy=(df.loc[low, "Roll_No"],
        df.loc[low, "Marks"]),
    xytext=(df.loc[low, "Roll_No"] + 1,
        df.loc[low, "Marks"] - 10),
    arrowprops=dict(facecolor="blue")
)

plt.title("Student Marks")

plt.grid(True)

plt.show()
```

```python
# ============================================================
# EXPERIMENT 3E : ADDING LEGENDS TO A GRAPH
# ============================================================
import pandas as pd
import matplotlib.pyplot as plt

# Read dataset
df = pd.read_csv("student_data.csv")
df["Roll_No"] = range(1, len(df) + 1)

# Create figure
plt.figure(figsize=(8,5))

# Plot Marks
plt.plot(
    df["Roll_No"],
    df["Marks"],
    label="Marks"
)

# Plot Attendance
plt.plot(
    df["Roll_No"],
    df["Attendance (%)"],
    label="Attendance"
)

# Add title
plt.title("Marks vs Attendance")

# Add axis labels
plt.xlabel("Roll Number")
plt.ylabel("Values")

# Display legend
plt.legend()

# Display grid
plt.grid(True)

# Display graph
plt.show()
```

```python
# ============================================================
# EXPERIMENT 3F : CREATING A SEABORN CATPLOT (FACTOR PLOT)
# ============================================================
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Read dataset
df = pd.read_csv("student_data.csv")
df["Roll_No"] = range(1, len(df) + 1)

# Create CatPlot
sns.catplot(
    data=df,
    x="Department",
    y="Marks",
    kind="bar"
)

# Add title
plt.title("Average Marks by Department")

# Display graph
plt.show()
```

```python
# ============================================================
# EXPERIMENT 3G : CREATING FIGURE-LEVEL PLOTS USING SEABORN RELPLOT
# ============================================================
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Read dataset
df = pd.read_csv("student_data.csv")


# Relationship Plot
sns.relplot(
    data=df,
    x="Study_Hours",
    y="Marks",
    kind="scatter",
    hue="Department"
)

plt.show()
```

```python
# ============================================================
# EXPERIMENT 3H : CREATING A DISTRIBUTION PLOT USING SEABORN DISPLOT
# ============================================================
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Read dataset
df = pd.read_csv("student_data.csv")

# Create distribution plot
sns.displot(
    data=df,
    x="Marks",
    bins=10,
    kde=True
)

# Display graph
plt.show()
```

```python
# ============================================================
# EXPERIMENT 3I : APPLYING COLOR MAPS USING MATPLOTLIB
# ============================================================
import pandas as pd
import matplotlib.pyplot as plt

# Read dataset
df = pd.read_csv("student_data.csv")
df["Roll_No"] = range(1, len(df) + 1)

# Correlation matrix
corr = df.corr(numeric_only=True)

# Create figure
plt.figure(figsize=(8, 6))

# Display color map
plt.imshow(
    corr,
    cmap="viridis",
    vmin=-1,
    vmax=1
)

# Show color scale
plt.colorbar(label="Correlation Coefficient")

# Axis tick labels
plt.xticks(
    range(len(corr.columns)),
    corr.columns,
    rotation=45,
    ha="right"
)

plt.yticks(
    range(len(corr.columns)),
    corr.columns
)

# Add correlation values inside each cell
for i in range(len(corr.columns)):
    for j in range(len(corr.columns)):
        plt.text(
            j, i,
            f"{corr.iloc[i, j]:.2f}",
            ha="center",
            va="center",
            color="white" if corr.iloc[i, j] < 0.5 else "black"
        )

# Add title
plt.title("Correlation Matrix using Color Map")

# Save the graph
plt.tight_layout()
plt.savefig("ColorMap.png")

# Display graph
plt.show()

print("Color map created successfully.")
```
