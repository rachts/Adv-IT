```python
# ============================================================
# EXPERIMENT 3A: ADDING CHART TITLE AND AXIS LABELS
# ============================================================
import pandas as pd
import matplotlib.pyplot as plt

# Read Dataset and Assign Roll Numbers
df = pd.read_csv("student_data.csv")
df["Roll_No"] = range(1, len(df) + 1)

# Create Figure and Plot Graph
plt.figure(figsize=(8, 5))
plt.plot(df["Roll_No"], df["Marks"], marker="o")

# Add Title and Axis Labels
plt.title("Student Marks Analysis")
plt.xlabel("Roll Number")
plt.ylabel("Marks")

# Display Graph
plt.show()
```

```python
# ============================================================
# EXPERIMENT 3B: ADDING GRID LINES TO A GRAPH
# ============================================================
import pandas as pd
import matplotlib.pyplot as plt

# Read Dataset and Assign Roll Numbers
df = pd.read_csv("student_data.csv")
df["Roll_No"] = range(1, len(df) + 1)

# Create Figure and Plot Attendance
plt.figure(figsize=(8, 5))
plt.plot(df["Roll_No"], df["Attendance (%)"], marker="o")

# Add Title, Labels, and Grid
plt.title("Attendance Analysis")
plt.xlabel("Roll Number")
plt.ylabel("Attendance (%)")
plt.grid(True)

# Display Graph
plt.show()
```

```python
# ============================================================
# EXPERIMENT 3C: ANNOTATING IMPORTANT DATA POINTS
# ============================================================
import pandas as pd
import matplotlib.pyplot as plt

# Read Dataset and Assign Roll Numbers
df = pd.read_csv("student_data.csv")
df["Roll_No"] = range(1, len(df) + 1)

# Create Figure and Plot Graph
plt.figure(figsize=(8, 5))
plt.plot(df["Roll_No"], df["Marks"], marker="o")

# Annotate Highest Marks
max_idx = df["Marks"].idxmax()
x, y = df.loc[max_idx, "Roll_No"], df.loc[max_idx, "Marks"]
plt.annotate("Highest Marks", xy=(x, y), xytext=(x + 2, y + 5), arrowprops=dict(facecolor="red"))

# Add Title and Grid
plt.title("Highest Marks")
plt.grid(True)

# Display Graph
plt.show()
```

```python
# ============================================================
# EXPERIMENT 3D: ADDING MULTIPLE ANNOTATIONS TO A GRAPH
# ============================================================
import pandas as pd
import matplotlib.pyplot as plt

# Read Dataset and Assign Roll Numbers
df = pd.read_csv("student_data.csv")
df["Roll_No"] = range(1, len(df) + 1)

# Create Figure and Plot Graph
plt.figure(figsize=(8, 5))
plt.plot(df["Roll_No"], df["Marks"], marker="o")

# Find Highest and Lowest Marks
high_idx, low_idx = df["Marks"].idxmax(), df["Marks"].idxmin()
hx, hy = df.loc[high_idx, "Roll_No"], df.loc[high_idx, "Marks"]
lx, ly = df.loc[low_idx, "Roll_No"], df.loc[low_idx, "Marks"]

# Add Annotations
plt.annotate("Highest", xy=(hx, hy), xytext=(hx + 1, hy + 5), arrowprops=dict(facecolor="green"))
plt.annotate("Lowest", xy=(lx, ly), xytext=(lx + 1, ly - 10), arrowprops=dict(facecolor="blue"))

# Add Title, Grid, and Display
plt.title("Student Marks")
plt.grid(True)
plt.show()
```

```python
# ============================================================
# EXPERIMENT 3E: ADDING LEGENDS TO A GRAPH
# ============================================================
import pandas as pd
import matplotlib.pyplot as plt

# Read Dataset and Assign Roll Numbers
df = pd.read_csv("student_data.csv")
df["Roll_No"] = range(1, len(df) + 1)

# Create Figure and Plot Data
plt.figure(figsize=(8, 5))
plt.plot(df["Roll_No"], df["Marks"], label="Marks")
plt.plot(df["Roll_No"], df["Attendance (%)"], label="Attendance")

# Add Title, Labels, Legend, and Grid
plt.title("Marks vs Attendance")
plt.xlabel("Roll Number")
plt.ylabel("Values")
plt.legend()
plt.grid(True)

# Display Graph
plt.show()
```

```python
# ============================================================
# EXPERIMENT 3F: CREATING A SEABORN CATPLOT (FACTOR PLOT)
# ============================================================
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Read Dataset and Assign Roll Numbers
df = pd.read_csv("student_data.csv")
df["Roll_No"] = range(1, len(df) + 1)

# Create CatPlot and Add Title
sns.catplot(data=df, x="Department", y="Marks", kind="bar")
plt.title("Average Marks by Department")

# Display Graph
plt.show()
```

```python
# ============================================================
# EXPERIMENT 3G: CREATING FIGURE-LEVEL PLOTS USING SEABORN RELPLOT
# ============================================================
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Read Dataset
df = pd.read_csv("student_data.csv")

# Create Relationship Plot and Display
sns.relplot(data=df, x="Study_Hours", y="Marks", kind="scatter", hue="Department")
plt.show()
```

```python
# ============================================================
# EXPERIMENT 3H: CREATING A DISTRIBUTION PLOT USING SEABORN DISPLOT
# ============================================================
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Read Dataset and Create Distribution Plot
df = pd.read_csv("student_data.csv")
sns.displot(data=df, x="Marks", bins=10, kde=True)

# Display Graph
plt.show()
```

```python
# ============================================================
# EXPERIMENT 3I: APPLYING COLOR MAPS USING MATPLOTLIB
# ============================================================
import pandas as pd
import matplotlib.pyplot as plt

# Read Dataset and Compute Correlation
df = pd.read_csv("student_data.csv")
df["Roll_No"] = range(1, len(df) + 1)
corr = df.corr(numeric_only=True)

# Create Figure and Display Color Map
plt.figure(figsize=(8, 6))
plt.imshow(corr, cmap="viridis", vmin=-1, vmax=1)
plt.colorbar(label="Correlation Coefficient")

# Configure Axis Ticks
cols = corr.columns
plt.xticks(range(len(cols)), cols, rotation=45, ha="right")
plt.yticks(range(len(cols)), cols)

# Add Correlation Values Inside Each Cell
for i in range(len(cols)):
    for j in range(len(cols)):
        plt.text(j, i, f"{corr.iloc[i, j]:.2f}", ha="center", va="center", color="white" if corr.iloc[i, j] < 0.5 else "black")

# Add Title, Save, and Display Graph
plt.title("Correlation Matrix using Color Map")
plt.tight_layout()
plt.savefig("ColorMap.png")
plt.show()

print("Color map created successfully.")
```
