```python
# ============================================================
# CREATE A SYNTHETIC STUDENT DATASET
# WITH MISSING VALUES AND DUPLICATE RECORDS
# ============================================================
import pandas as pd
import numpy as np

# Set random seed for reproducibility
np.random.seed(42)

# Number of students
n = 20

# Create synthetic dataset
student_data = pd.DataFrame({
    "Roll_No": np.arange(101, 101 + n),
    "Name": [f"Student_{i}" for i in range(1, n + 1)],
    "Age": np.random.randint(18, 24, n),
    "Gender": np.random.choice(["Male", "Female"], n),
    "Department": np.random.choice(["CSE", "IT", "ECE", "EE", "ME"], n),
    "Semester": np.random.randint(1, 9, n),
    "Attendance (%)": np.random.randint(60, 101, n),
    "Marks": np.random.randint(35, 101, n),
    "Study_Hours": np.round(np.random.uniform(1, 8, n), 1),
    "Sports": np.random.choice(["Yes", "No"], n)
})

# --------------------------------------------
# Introduce Missing Values
# --------------------------------------------

student_data.loc[2, "Age"] = np.nan
student_data.loc[5, "Marks"] = np.nan
student_data.loc[7, "Department"] = np.nan
student_data.loc[10, "Attendance (%)"] = np.nan
student_data.loc[14, "Study_Hours"] = np.nan
student_data.loc[18, "Gender"] = np.nan

# --------------------------------------------
# Introduce Duplicate Records
# --------------------------------------------

student_data = pd.concat(
    [student_data, student_data.iloc[[3]], student_data.iloc[[8]]],
    ignore_index=True
)

# Save the dataset
student_data.to_csv("student_data_dirty.csv", index=False)

# Display dataset
print("Dataset with Missing Values and Duplicate Records")
print(student_data)

print("\nDataset Shape :", student_data.shape)

print("\nDataset saved as 'student_data_dirty.csv'")
```

```python
# ============================================================
# EXPERIMENT 1b : DATA STATISTICS USING PANDAS
# ============================================================

# Import the Pandas library
import pandas as pd

# Create the dataset file (run once)
import pandas as pd

data = pd.DataFrame({
    "Name": ["Amit", "Priya", "Rahul", "Sneha", "Karan", "Anita", "Vikram", "Pooja"],
    "Department": ["CSE", "IT", "CSE", "ECE", "IT", "CSE", "ECE", "IT"],
    "Marks": [78, 85, 67, 90, 72, 88, 60, 95],
    "Attendance (%)": [85, 90, 75, 95, 80, 92, 70, 98]
})

data.to_csv("student_data.csv", index=False)
print("student_data.csv created successfully")
# Read the dataset
data = pd.read_csv("student_data.csv")

# Display the first five records
print("========== FIRST FIVE RECORDS ==========")
print(data.head())

# Display descriptive statistics for all numerical columns
print("\n========== DESCRIPTIVE STATISTICS ==========")
print(data.describe())

# Calculate the mean of Marks
print("\nAverage Marks")
print(data["Marks"].mean())

# Calculate the median of Marks
print("\nMedian Marks")
print(data["Marks"].median())

# Calculate the mode of Department
print("\nMost Common Department")
print(data["Department"].mode())

# Calculate minimum marks
print("\nMinimum Marks")
print(data["Marks"].min())

# Calculate maximum marks
print("\nMaximum Marks")
print(data["Marks"].max())

# Calculate standard deviation
print("\nStandard Deviation of Marks")
print(data["Marks"].std())

# Calculate variance
print("\nVariance of Marks")
print(data["Marks"].var())

# Calculate average attendance
print("\nAverage Attendance (%)")
print(data["Attendance (%)"].mean())

# Display correlation among numerical columns
print("\nCORRELATION MATRIX")
print(data.corr(numeric_only=True))
```
