```python
# ============================================================
# CREATE A SYNTHETIC STUDENT DATASET WITH MISSING VALUES AND DUPLICATES
# ============================================================
import pandas as pd
import numpy as np

# Generate Synthetic Dataset
np.random.seed(42)
n = 20
student_data = pd.DataFrame({
    "Roll_No": np.arange(101, 101 + n),
    "Name": [f"Student_{i}" for i in range(1, n + 1)],
    "Age": np.random.randint(18, 24, n),
    "Gender": np.random.choice(["Male", "Female"], n),
    "Department": np.random.choice(["CSE", "IT", "ECE", "EE", "ME"], n),
    "Semester": np.random.randint(1, 9, n),
    "Attendance (%)": np.random.randint(60, 101, n),
    "Marks": np.random.randint(35, 101, n),
    "Study_Hours": np.round(np.random.uniform(1, 8, n), 1),
    "Sports": np.random.choice(["Yes", "No"], n)
})

# Introduce Missing Values
for idx, col in [(2, "Age"), (5, "Marks"), (7, "Department"), (10, "Attendance (%)"), (14, "Study_Hours"), (18, "Gender")]:
    student_data.loc[idx, col] = np.nan

# Introduce Duplicate Records
student_data = pd.concat([student_data, student_data.iloc[[3, 8]]], ignore_index=True)

# Save and Display Dataset
student_data.to_csv("student_data_dirty.csv", index=False)
print(f"Dataset with Missing Values and Duplicate Records\n{student_data}")
print(f"\nDataset Shape : {student_data.shape}")
print("\nDataset saved as 'student_data_dirty.csv'")
```

```python
# ============================================================
# EXPERIMENT 1B: DATA STATISTICS USING PANDAS
# ============================================================
import pandas as pd

# Create and Save Dataset
data = pd.DataFrame({
    "Name": ["Amit", "Priya", "Rahul", "Sneha", "Karan", "Anita", "Vikram", "Pooja"],
    "Department": ["CSE", "IT", "CSE", "ECE", "IT", "CSE", "ECE", "IT"],
    "Marks": [78, 85, 67, 90, 72, 88, 60, 95],
    "Attendance (%)": [85, 90, 75, 95, 80, 92, 70, 98]
})
data.to_csv("student_data.csv", index=False)
print("student_data.csv created successfully")

# Read the Dataset
data = pd.read_csv("student_data.csv")

# Heading: First Five Records
print(f"========== FIRST FIVE RECORDS ==========\n{data.head()}")

# Heading: Descriptive Statistics
print(f"\n========== DESCRIPTIVE STATISTICS ==========\n{data.describe()}")

# Heading: Average Marks
print(f"\nAverage Marks\n{data['Marks'].mean()}")

# Heading: Median Marks
print(f"\nMedian Marks\n{data['Marks'].median()}")

# Heading: Most Common Department
print(f"\nMost Common Department\n{data['Department'].mode()}")

# Heading: Minimum Marks
print(f"\nMinimum Marks\n{data['Marks'].min()}")

# Heading: Maximum Marks
print(f"\nMaximum Marks\n{data['Marks'].max()}")

# Heading: Standard Deviation of Marks
print(f"\nStandard Deviation of Marks\n{data['Marks'].std()}")

# Heading: Variance of Marks
print(f"\nVariance of Marks\n{data['Marks'].var()}")

# Heading: Average Attendance (%)
print(f"\nAverage Attendance (%)\n{data['Attendance (%)'].mean()}")

# Heading: Correlation Matrix
print(f"\nCORRELATION MATRIX\n{data.corr(numeric_only=True)}")
```
