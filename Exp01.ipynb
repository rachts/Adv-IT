```python
# ============================================================
# EXPERIMENT 1C: DATA CLEANING USING PANDAS
# ============================================================
import pandas as pd

data = pd.read_csv("student_data_dirty.csv")

# Heading: Original Dataset
print(f"========== ORIGINAL DATASET ==========\n{data.head()}")

# Heading: Missing Values
print(f"\n========== MISSING VALUES ==========\n{data.isnull()}")

# Heading: Total Missing Values
print(f"\n========== TOTAL MISSING VALUES ==========\n{data.isnull().sum()}")

# Heading: Duplicate Records
print(f"\n========== DUPLICATE RECORDS ==========\n{data.duplicated().sum()}")

# Remove Duplicates, Fill Numerical (Mean) and Categorical (Mode) Values
data = data.drop_duplicates().copy()
for c in ["Marks", "Attendance (%)", "Study_Hours"]:
    data[c] = data[c].fillna(data[c].mean())
for c in ["Department", "Gender", "Sports"]:
    data[c] = data[c].fillna(data[c].mode()[0])

# Rename Column and Convert Data Type
data.rename(columns={"Attendance (%)": "Attendance"}, inplace=True)
data["Semester"] = data["Semester"].astype(int)

# Heading: Cleaned Dataset
print(f"\n========== CLEANED DATASET ==========\n{data.head()}")

# Heading: Dataset Information
print("\n========== DATASET INFORMATION ==========")
data.info()

# Save the Cleaned Dataset
data.to_csv("student_data_cleaned.csv", index=False)
print("\nCleaned dataset has been saved as student_data_cleaned.csv")
```

```python
# ============================================================
# EXPERIMENT 1D: BAR CHART USING MATPLOTLIB
# ============================================================
import pandas as pd
import matplotlib.pyplot as plt

# Read Dataset and Compute Department-wise Average Marks
data = pd.read_csv("student_data.csv")
average_marks = data.groupby("Department")["Marks"].mean()
print(f"Average Marks of Each Department\n{average_marks}")

# Create and Configure Bar Chart
plt.figure(figsize=(8, 5))
plt.bar(average_marks.index, average_marks.values, color=['red', 'blue', 'green', 'orange', 'purple'])
plt.title("Average Marks of Students by Department")
plt.xlabel("Department"), plt.ylabel("Average Marks")

# Display Values on Top of Each Bar
for i, value in enumerate(average_marks.values):
    plt.text(i, value + 0.5, f"{value:.1f}", ha='center')

plt.grid(axis='y', linestyle='--', alpha=0.5), plt.show()
```

```python
# ============================================================
# EXPERIMENT 1E: PIE CHART USING MATPLOTLIB
# ============================================================
import pandas as pd
import matplotlib.pyplot as plt

# Read Dataset and Count Department-wise Students
data = pd.read_csv("student_data.csv")
department_count = data["Department"].value_counts()
print(f"Number of Students in Each Department\n{department_count}")

# Create and Configure Pie Chart
plt.figure(figsize=(8, 8))
plt.pie(department_count.values, labels=department_count.index, autopct='%1.1f%%', startangle=90, shadow=True)
plt.title("Distribution of Students by Department")
plt.show()
```

```python
# ============================================================
# EXPERIMENT 1G: BOX PLOT USING MATPLOTLIB
# ============================================================
import pandas as pd
import matplotlib.pyplot as plt

# Read Dataset and Display Records and Statistics
data = pd.read_csv("student_data.csv")
print(f"First Five Records\n{data.head()}")
print(f"\nStatistics of Marks\n{data['Marks'].describe()}")

# Create and Configure Box Plot
plt.figure(figsize=(6, 6))
plt.boxplot(data["Marks"], patch_artist=True, labels=["Marks"],
            boxprops=dict(facecolor="lightgreen"),
            medianprops=dict(color="red", linewidth=2),
            whiskerprops=dict(color="blue"),
            capprops=dict(color="black"),
            flierprops=dict(marker='o', markerfacecolor='red', markersize=8))
plt.title("Box Plot of Student Marks"), plt.ylabel("Marks")
plt.grid(axis='y', linestyle='--', alpha=0.5), plt.show()
```

```python
# ============================================================
# EXPERIMENT 1I: TRIANGULATION PLOT (adapted to 4-column dataset)
# ============================================================
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.tri as tri

data = pd.read_csv("student_data.csv")
print(f"First Five Records\n{data.head()}")
x, y = data["Attendance (%)"], data["Marks"]   # use Attendance instead of Study_Hours

triangulation = tri.Triangulation(x, y)
plt.figure(figsize=(8, 6))
plt.triplot(triangulation, color='blue')
plt.scatter(x, y, color='red')

plt.title("Triangulation Plot of Attendance vs Marks")
plt.xlabel("Attendance (%)"), plt.ylabel("Marks")
plt.grid(True), plt.show()
```
